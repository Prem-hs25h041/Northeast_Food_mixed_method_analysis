#### Research Questions : 

1. Why is the consumption of ethnic food so crucial for the Northeast migrants, despite

2. In what ways do the ethnic store visuals, images, and symbols evoke expressions of the

communities?

the local economy, including employment generation, entrepreneurship and income

economies, and enhancing food security and nutrition?

```python
def make_columns_unique(df):
    seen = {}
    new_columns = []

    for col in df.columns:
        if col not in seen:
            seen[col] = 0
            new_columns.append(col)
        else:
            seen[col] += 1
            new_columns.append(f"{col}_{seen[col]}")

    df = df.copy()
    df.columns = new_columns
    return df



df1 = make_columns_unique(df1)
df2 = make_columns_unique(df2)



print("df1 columns unique:", df1.columns.is_unique)
print("df2 columns unique:", df2.columns.is_unique)



df = pd.concat([df1, df2], ignore_index=True)

print("Final shape:", df.shape)
```

```python
### Reusable Bar plot function - eg : plot_bar(quant_df,'gender')

def plot_bar(quant_df, column, top_n=None):
    counts = quant_df[column].value_counts(dropna=False)

    if top_n:
        counts = counts.head(top_n)

    plt.figure(figsize=(10, 5))
    counts.plot(kind='bar')

    plt.title(f"Distribution of {column}")
    plt.xlabel(column)
    plt.ylabel("Number of Respondents")
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.show()
```

```python
#### Reusable Pie Chart function - eg : plot_pie(quant_df,"gender")

def plot_pie(quant_df, column):
    counts = quant_df[column].value_counts(dropna=False)

    plt.figure(figsize=(7, 7))

    plt.pie(
        counts,
        labels=counts.index,
        autopct='%1.1f%%',
        startangle=90
    )

    plt.title(f"Distribution of {column}")
    plt.tight_layout()
    plt.show()
```

## Conclusion 

#### Ethnic food is much more than food for Northeast migrants—it is strongly tied to cultural identity.

87.77% agree/strongly agree that ethnic food keeps them connected to their culture.

84.39% consider ethnic food important/very important for maintaining cultural identity.

56.54% say ethnic food directly strengthens their sense of identity.

91.56% consider ethnic food important/very important during celebrations and festivals.

#### Cultural attachment remains strong even when discrimination is perceived.

41.77% of respondents believe Northeast people face discrimination/hatred associated with the emergence of ethnic food stores in neighbourhoods, while 54.43% disagree.

Importantly, among respondents who do perceive discrimination:

72.72% still consider ethnic food important/very important.

91.91% agree/strongly agree that ethnic food connects them to their culture.

61.62% say it strengthens their identity.

84.85% say ethnic stores induce pride in their cultural heritage.

Thus, the analysis suggests that cultural and identity-related benefits of ethnic food remain powerful despite perceived social tensions.

#### Ethnic food stores act as important community spaces.

84.39% agree/strongly agree that ethnic stores create a sense of belonging.

56.54% consider them social gathering places, while another 37.55% say they sometimes perform this role.

73.84% say ethnic food stores contribute to social cohesion and community building.

Among people who frequently attend ethnic-food-centred gatherings, 81.08% believe these stores contribute to social cohesion.

This indicates that ethnic food spaces function as social and community infrastructure for Northeast migrants, not merely as commercial outlets.

#### Sharing ethnic food also appears to facilitate social interaction.

54.01% sometimes share ethnic food with friends/colleagues and 26.16% do so often.

Among those who share ethnic food often, 85.48% believe ethnic food stores contribute to community cohesion.

Ethnic food can therefore act as a medium for interaction, cultural exchange and community formation.

#### Ethnic food stores are the dominant source for obtaining traditional ingredients.

77.22% purchase ingredients from ethnic food stores.

Local markets are used by about 25.32%.

64.56% discover new ethnic stores through word of mouth from friends, much higher than social media (28.69%) or online search (7.59%).

This again highlights the importance of community networks in sustaining Northeast food consumption.

#### Access to ethnic ingredients remains a significant problem.

27.43% say they face sourcing difficulties and another 44.30% say they sometimes face them—together about 71.73%.

Major reported problems include:

High prices – 39.24%

Availability of ingredients – 36.71%

Quality of ingredients – 23.63%

Distance/access to stores is also repeatedly reported.

Hence, migrants maintain their ethnic food practices despite noticeable cost and availability constraints.

#### Ethnic food stores are perceived as improving quality of life and well-being.

77.64% agree/strongly agree that the presence of ethnic food stores improves quality of life.

88.61% agree/strongly agree that ethnic food stores and ethnic food consumption contribute to the well-being of the Northeast community.

81.01% consider ethnic food stores important/very important for access to safe, nutritious and sufficient food.

#### Respondents strongly perceive economic benefits from ethnic food stores.

83.12% agree/strongly agree that ethnic food stores contribute to the local economy and create opportunities for the Northeast community.

79.32% agree/strongly agree that they support gender equality by creating opportunities for women entrepreneurs.

Therefore, respondents see these businesses as combining cultural preservation with entrepreneurship and economic opportunity.

#### Ethnic food stores also contribute to the multicultural character of metropolitan cities.

63.71% identify cultural exchange as a contribution.

50.21% identify increased culinary diversity.

Respondents also identify social cohesion and economic opportunities.

Thus, these spaces potentially connect Northeast culture with the broader urban population, rather than functioning solely within migrant communities.

#### The environmental sustainability dimension is much weaker.

Only 14.35% believe ethnic food stores engage in environmentally sustainable practices.

53.59% don't know, while 30.38% say no.

Therefore, the notebook provides good support for the cultural, social, economic and food-security dimensions of the SDGs, but not strong evidence for environmental sustainability.

#### Overall finding

The strongest conclusion from the analysis is that ethnic food stores and ethnic food consumption function as a form of cultural and social infrastructure for Northeast migrants in metropolitan cities.

They help migrants preserve identity, maintain cultural practices, develop a sense of belonging, build community networks, access traditional food and improve perceived well-being.

At the same time, ethnic food stores are perceived to generate economic and entrepreneurial opportunities, particularly including opportunities for women.

This cultural attachment remains strong even among respondents who perceive discrimination or hostility, suggesting that ethnic food can serve as an important mechanism of cultural resilience in migrant life.

The main obstacles are high prices, ingredient availability, quality and accessibility of stores.

#### Limitations 

The sample is not evenly distributed:

Delhi: 61.60%

Chennai: 27.00%

Bangalore: 6.33%

64.14% are aged 18–25.

52.74% are students.

43.88% are from Manipur.

Therefore, the findings should not automatically be generalized to all Northeast migrants in Indian metropolitan cities.

The notebook currently uses mainly frequencies, percentages and cross-tabulations. It does not conduct statistical significance tests such as Chi-square tests or regression, so relationships should be described as associations/patterns rather than causal effects.

Research Question 2 about store visuals, images and symbols is not substantively analysed in the notebook.

RQ4 asks about actual employment generation, entrepreneurship and income generation, but the notebook mainly measures respondents' perceptions of economic contribution. It therefore cannot yet establish the actual number of jobs, businesses or income generated.

Similarly, RQ1 refers to violence and conflict, whereas the available quantitative variable measures perceived discrimination/hatred, not actual incidents of violence. So the conclusion should not claim that the dataset directly measures violence.

In one sentence: The data strongly indicate that Northeast ethnic food and ethnic food stores serve as important mechanisms of cultural preservation, identity, belonging, community cohesion, food security and perceived economic opportunity for Northeast migrants, even in the presence of sourcing difficulties and perceived social discrimination, although stronger statistical testing and more direct economic/qualitative evidence are needed for causal or generalizable claims.
